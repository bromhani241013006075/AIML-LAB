```python
from collections import deque

def water_jug():
    start = (0, 0)

    q = deque([(start, [])])
    visited = {start}

    while q:
        (a, b), path = q.popleft()

        # Check if we have 2 liters
        if a == 2 or b == 2:
            return path + [(a, b)]

        # Possible states
        states = [
            (4, b),  # Fill 4-liter jug
            (a, 3),  # Fill 3-liter jug
            (0, b),  # Empty 4-liter jug
            (a, 0),  # Empty 3-liter jug

            # Pour 4-liter jug into 3-liter jug
            (
                a - min(a, 3 - b),
                b + min(a, 3 - b)
            ),

            # Pour 3-liter jug into 4-liter jug
            (
                a + min(b, 4 - a),
                b - min(b, 4 - a)
            )
        ]

        # Add new states to the queue
        for state in states:
            if state not in visited:
                visited.add(state)
                q.append((state, path + [state]))


# Find the solution
solution = water_jug()

# Print the solution step by step
for i, state in enumerate(solution):
    print(i, state)
```
