```python
# Graph representing cities and routes
graph = {
    'A': ['B'],
    'B': ['A', 'C'],
    'C': ['B'],
    'D': ['E'],
    'E': ['D', 'F'],
    'F': ['E']
}


# Function to check if a route exists between two cities
def route_exists(graph, start, target, visited=None):
    if visited is None:
        visited = set()

    if start == target:
        return True

    visited.add(start)

    for city in graph[start]:
        if city not in visited:
            if route_exists(graph, city, target, visited):
                return True

    return False


# Function to find connected components
def connected_components(graph):
    visited = set()
    components = []

    def dfs(city, component):
        visited.add(city)
        component.append(city)

        for nxt in graph[city]:
            if nxt not in visited:
                dfs(nxt, component)

    for city in graph:
        if city not in visited:
            component = []
            dfs(city, component)
            components.append(component)

    return components


# Print results
print("Route A to C:", route_exists(graph, 'A', 'C'))
print("Route A to F:", route_exists(graph, 'A', 'F'))
print("Connected components:", connected_components(graph))
```
